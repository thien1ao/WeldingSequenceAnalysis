```python
import pandas as pd
import polars as pl
import glob

import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

import numpy as np
import math
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.neighbors import NearestNeighbors

from sklearn.cluster import AgglomerativeClustering, DBSCAN
from sklearn.metrics import adjusted_rand_score, silhouette_score

from scipy.ndimage import minimum_filter1d, maximum_filter1d, uniform_filter

from kneed import KneeLocator

import ipywidgets as widgets
from IPython.display import display, clear_output

from scipy.signal import spectrogram

import settings_

import warnings
warnings.filterwarnings('ignore')
```

```python
files = sorted(glob.glob('./data/Daten/EV__*.parquet'))

lf = pl.scan_parquet(files)
schema = lf.collect_schema()

lf.head(10).collect()
```

```python
EIN = "[72_1]S7-Schweißmaschine - Schweißen EIN (M215_0)"

features = [col for col, dtype in schema.items() if col not in ("Time", EIN) and dtype.is_numeric()]

df_features = lf.select("Time", EIN, *features).drop_nulls(subset=["Time", EIN])
df_features = df_features.with_columns(seq_id=(pl.col(EIN) & ~pl.col(EIN).shift(fill_value=False)).cum_sum() - 1)
df_features = df_features.filter(pl.col(EIN)).select("Time", "seq_id", *features)

df_features.slice(18190, 10).collect(engine="streaming")
```

```python
means = df_features.select([pl.col(col).mean().alias(col) for col in features]).collect(engine="streaming")
stds = df_features.select([pl.col(col).std().alias(col) for col in features]).collect(engine="streaming")

means = np.array([means[col][0] for col in features])
stds = np.array([stds[col][0] for col in features])

means_np = np.asarray(means, dtype=np.float64)
stds_np = np.asarray(stds, dtype=np.float64)

summary = pl.DataFrame({
    "feature": features,
    "mean": means_np,
    "std": stds_np
})

const_features = (
    summary
    .filter(pl.col("std") == 0)
    .get_column("feature")
    .to_list()
    
)

valid = [i for i, col in enumerate(features) if col not in const_features]
means_np = means_np[valid]
stds_np = stds_np[valid]
features = [features[i] for i in valid]

df_features = lf.select("Time", EIN, *features).drop_nulls(subset=["Time", EIN])
df_features = df_features.with_columns(seq_id=(pl.col(EIN) & ~pl.col(EIN).shift(fill_value=False)).cum_sum() - 1)
df_features = df_features.filter(pl.col(EIN)).select("Time", "seq_id", *features)
```

```python
df_seq = (
    df_features
    .group_by("seq_id")
    .agg(start=pl.col("Time").min(), end=pl.col("Time").max())
    .sort("seq_id")
    .collect(engine="streaming")
)

def interpolate(N):
    X = np.empty((len(df_seq), len(features) * N), dtype=np.float64)
    seq_ids = df_seq["seq_id"].to_list()
    norm = [((pl.col(col).cast(pl.Float64) - mean) / std).alias(col) for col, mean, std in zip(features, means_np, stds_np)]
    empty = []

    for i, row in enumerate(df_seq.iter_rows(named=True)):
        seq = lf.filter(pl.col("Time").is_between(row["start"], row["end"]))
        seq = seq.filter(pl.col(EIN))
        seq = seq.select("Time", *features)
        seq = seq.sort("Time")
        seq = seq.with_columns(norm)
        seq = seq.collect(engine="streaming")

        t = (seq["Time"] - seq["Time"][0]).dt.total_milliseconds().to_numpy()
        t_lin = np.linspace(0, t[-1], N)
        
        for j, feature in enumerate(features):
            y = seq[feature].to_numpy()
            
            dummy = np.isfinite(y)
            if not dummy.any():
                empty.append((row["seq_id"], feature))
                X[i, j*N:(j+1)*N] = 0.0
                continue
            X[i, j*N:(j+1)*N] = np.interp(t_lin, t[dummy], y[dummy])
    return X, seq_ids, empty

def build_df_X(N):
    X, seq_ids, _ = interpolate(N)

    cols = [f"{feature}_t_{i}" for feature in features for i in range(N)]

    return pd.DataFrame(X, index=pd.Index(seq_ids, name="seq_id"), columns=cols)
```

```python
df_X = build_df_X(settings_.N_SAMPLES)
X = df_X.to_numpy()
df_X.iloc[:10, ::settings_.N_SAMPLES] # 10 rows
```

```python
pca2 = PCA(n_components=2)

df_pca = pd.DataFrame(pca2.fit_transform(X), columns=["PC1", "PC2"], index=df_X.index)

plt.figure(figsize=(16, 5))
plt.subplot(121)
plt.scatter(df_pca["PC1"], df_pca["PC2"], s=16, alpha=0.6, color="darkorchid")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(alpha=0.4)

plt.subplot(122)
PCA_ = PCA()
PCA_.fit(X)

cum_var = PCA_.explained_variance_ratio_.cumsum()

plt.plot(range(1, len(cum_var) + 1), cum_var, color="darkorchid")

plt.axhline(y=0.95, linestyle="--", label="95% var", color="orange")
plt.grid(True)
plt.legend()
plt.show()
```

```python
var = PCA_.explained_variance_ratio_

def get_components(var):
    total = 0
    n = 0

    for v in var:
        total += v
        n += 1

        if  total >= 0.95:
            break
    return n
        
X_PCA = PCA_.transform(X)
X_reduced = X_PCA[:, :get_components(var)]

def elbow(n_init, k_max=10, data=X_reduced):
    ks = np.arange(1,k_max+1)
    inertias = [KMeans(n_clusters=k, n_init=n_init, random_state=0).fit(data).inertia_ for k in ks]
    
    return ks, inertias

ks, inertias = elbow(n_init=50)
ret = np.array(inertias)

sil_ks = ks[1:]
sil = [silhouette_score(X_reduced, KMeans(n_clusters=k, n_init=50, random_state=0).fit_predict(X_reduced)) for k in sil_ks]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(ks, ret, marker="o", color="darkorchid")
ax1.set_title("Elbow")
ax1.set_xlabel("k")
ax1.set_ylabel("inertia")
ax1.grid(alpha=0.4)

ax2.plot(sil_ks, sil, marker="o", color="darkorange")
ax2.set_title("Silhouette")
ax2.set_xlabel("k")
ax2.set_ylabel("silhouette score")
ax2.grid(alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
cluster_containers = []

for k in [2, 3, 4]:
    tmp = df_pca[["PC1", "PC2"]].copy()
    tmp["cluster"] = KMeans(n_clusters=k, n_init=50, random_state=0).fit_predict(X_reduced).astype(str)
    tmp["k"] = k
    cluster_containers.append(tmp)

main_container = pd.concat(cluster_containers)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), sharex=True, sharey=True)

for ax, k in zip(axes, [2, 3, 4]):
    k_container = main_container[main_container["k"] == k]

    for cluster in k_container["cluster"]:
        cluster_data = k_container[k_container["cluster"] == cluster]
        ax.scatter(cluster_data["PC1"], cluster_data["PC2"], s=10, alpha=0.7)

    ax.set_title(f"k = {k}")
    ax.set_xlabel("PCA1")
    ax.set_ylabel("PCA2")

plt.tight_layout()
plt.show()
```

```python
palette = settings_.palette
eps_list = settings_.eps_list

ne, nm = len(eps_list), len(min_samples)
pca_c = df_pca.to_numpy()

fig, axes = plt.subplots(ne, nm, figsize=(3 * nm, 2.6 * ne), sharex=True, sharey=True, squeeze=False, constrained_layout=True)

history = {"eps": [], "min_samples": [], "n_clusters": [], "noise": [], "silhouette": []}
for i, e in enumerate(eps_list):
    for j, ms in enumerate(min_samples):
        ax = axes[i, j]
        db = DBSCAN(eps=e, min_samples=ms).fit_predict(X_reduced)
        mask = db != -1
        n_cl = len(set(db[mask]))
        noise = (~mask).mean()
        sil = silhouette_score(X_reduced[mask], db[mask]) if n_cl >= 2 else np.nan
        
        history["eps"].append(e)
        history["min_samples"].append(ms)
        history["n_clusters"].append(n_cl)
        history["noise"].append(noise)
        history["silhouette"].append(sil)

        ax.scatter(*pca_c[~mask].T, c="lightgrey", s=4, alpha=0.6, linewidths=0)
        colors = [palette[k % len(palette)] for k in db[mask]]
        ax.scatter(*pca_c[mask].T, c=colors, s=6, alpha=0.85, linewidths=0)

        sil_txt = f"{sil:.2f}" if not np.isnan(sil) else "–"
        ax.text(0.02, 0.98, f"k={n_cl}  noise={noise:.0%}  sil={sil_txt}",
                transform=ax.transAxes, va="top", fontsize=8,
                bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="none", alpha=0.8))
        ax.tick_params(labelsize=6)

        if i == 0:      
            ax.set_title(f"min_samples = {ms}", fontsize=10, fontweight="bold")
        if j == 0:      
            ax.set_ylabel(f"eps = {e}", fontsize=10, fontweight="bold")

fig.supxlabel("PC1")
fig.supylabel("PC2", x=-0.01)
plt.show()

df_history = pd.DataFrame(history)
```

```python
N_CLUSTERS = settings_.N_CLUSTERS
MAX_NOISE = settings_.MAX_NOISE
N_BEST = settings_.N_BEST

piv = df_history.pivot(index="eps", columns="min_samples", values="n_clusters")

def stability(eps, ms):
    i, j = piv.index.get_loc(eps), piv.columns.get_loc(ms)
    kernel = piv.values[max(i-1, 0): i+2, max(j-1, 0): j+2]
    return (kernel == piv.values[i, j]).mean()

df_history["stability"] = [stability(e, ms) for e, ms in zip(df_history["eps"], df_history["min_samples"])]
candidates = df_history[(df_history["n_clusters"] == N_CLUSTERS) & (df_history["noise"] <= MAX_NOISE)]
best = candidates.sort_values(["stability", "silhouette"], ascending=False).head(N_BEST)

the_ones = [(int(e), int(ms)) for e, ms in zip(best["eps"], best["min_samples"])]
```

```python
N_SEQ = settings_.N_SEQ
e, ms = the_ones[0]
db = DBSCAN(eps=e, min_samples=ms).fit_predict(X_reduced)
CHOSEN_CLUSTER = settings_.CHOSEN_CLUSTER

def get_seq(n_seq, e, ms, chosen_cluster):
    centers = np.stack([X_reduced[km3 == c].mean(axis=0) for c in sorted(set(km3))])
    dist_c = np.linalg.norm(X_reduced - centers[km3], axis=1)
    nearest_seq = {c: np.where(km3 == c)[0][dist_c[km3 == c].argmin()] for c in sorted(set(km3))}
    noise_pos = np.where(db == -1)[0]
    outliers = noise_pos[np.argsort(dist_c[noise_pos])[::-1][:N_SEQ]]

    cluster_indices = np.where(km3 == chosen_cluster)[0]
    cluster_dist = dist_c[cluster_indices]

    nearest_pos = cluster_dist.argmin()
    farthest_pos = cluster_dist.argmax()

    nearest_in_cluster = cluster_indices[nearest_pos]
    farthest_in_cluster = cluster_indices[farthest_pos]
    cluster_pair = [nearest_in_cluster, farthest_in_cluster]

    return nearest_seq, outliers, cluster_pair
    
def load_uniterpolated(seq_id):
    bounds = df_seq.row(seq_id, named=True)
    start, end = bounds["start"], bounds["end"]

    in_segment = pl.col("Time").is_between(start, end)
    is_active = pl.col(EIN)

    seq_data = lf.filter(in_segment & is_active).select("Time", *features).sort("Time").collect(engine="streaming")
    t = (seq_data["Time"] - seq_data["Time"][0]).dt.total_milliseconds().to_numpy() / 1000

    return t, seq_data

nearest_seq, outliers, cluster_pair = get_seq(N_SEQ, e, ms, CHOSEN_CLUSTER)
```

```python
def build_curves(df_seq, nearest_seq, cluster_pair, outliers, km3, chosen_cluster, styles=settings_.STYLES):
    duration = (df_seq["end"] - df_seq["start"]).dt.total_milliseconds().to_numpy() / 1000
    curves = []
    pn, pf = cluster_pair[0], cluster_pair[1]

    for c, seq_id in nearest_seq.items():
        median = np.median(duration[km3 == c])
        res =f"Cluster {c} (Median {median:.1f} s)"
        curves.append((res, styles.cluster_colors[c], "-", load_uniterpolated(seq_id)))

    curves.append((f"Cluster {chosen_cluster} (am weitesten)", styles.chosen_color, ":", load_uniterpolated(pf)))

    for i, (seq_id, (color, ls)) in enumerate(zip(outliers, styles.outlier_styles)):
        curves.append((f"Outlier {i}", color, ls, load_uniterpolated(seq_id)))

    return curves


def smoothing(signal, smooth_size):
    return uniform_filter(signal, size=smooth_size, mode="nearest")


def plot_features(features, curves, ncols=3, per_fig=12, **kwargs):
    for i in range(0, len(features), per_fig):
        chunk = features[i:i + per_fig]
        nrows = math.ceil(len(chunk) / ncols)

        fig, axes = plt.subplots(nrows, ncols, figsize=(5 * ncols, 3.5 * nrows), squeeze=False)
        axes = axes.ravel()
        for ax, feature in zip(axes, chunk):
            plot_feature(feature, ax, curves, **kwargs)

        for ax in axes[len(chunk):]:
            ax.axis("off")
            
        axes[0].legend(fontsize=7, loc="upper right")
        plt.tight_layout()
        plt.show()

        
def plot_feature(feature, ax, curves, suffix="", envelope=False, time_range=None, 
                 extras=True, smooth_size=None, envelope_size=settings_.ENVELOPE_SIZE):
    for cl_n, color, ls, (t, seq) in curves:
        if ls != "-" and not extras:
            continue
            
        y = seq[feature].to_numpy().astype(float)

        if smooth_size:
            y = smoothing(y, smooth_size)
        if envelope:
            lower = minimum_filter1d(y, envelope_size)
            upper = maximum_filter1d(y, envelope_size)

        if time_range:
            mask = (t >= time_range[0]) & (t <= time_range[1])
            t, y = t[mask], y[mask]
            if envelope:
                lower = lower[mask]
                upper = upper[mask]
                
        if envelope:
            if ls == "-":
                ax.fill_between(t, lower, upper, color=color, alpha=0.25, lw=0)
                ax.plot(t, upper, color=color, lw=1, label=cl_n)
                ax.plot(t, lower, color=color, lw=1)
            else:
                ax.plot(t, (lower + upper) / 2, color=color, ls=ls, lw=1, label=cl_n)
        else:
            ax.plot(t, y, color=color, ls=ls, lw=1, label=cl_n)
            
    ax.set_title(feature + suffix, fontsize=10)
    ax.set_xlabel("t")
    ax.grid(alpha=0.4)

def feature_viewer(features, curves):
    feature = widgets.Dropdown(options=features, description="Feature:", layout=widgets.Layout(width="400px"))
    envelope = widgets.Checkbox(value=False, description="Envelope")
    extras = widgets.Checkbox(value=True, description="Outliers / extras")
    smooth = widgets.IntSlider(value=0, min=0, max=200, step=5, description="Smoothing:", continuous_update=False)

    def update(feature, envelope, extras, smooth):
        fig, ax = plt.subplots(figsize=(11, 5))
        plot_feature(feature, ax, curves, envelope=envelope, extras=extras, smooth_size=smooth if smooth > 0 else None)

        ax.legend(fontsize=8, ncol=3)
        plt.tight_layout()
        plt.show()
        plt.close(fig)

    plot = widgets.interactive_output(
        update,
        {
            "feature": feature,
            "envelope": envelope,
            "extras": extras,
            "smooth": smooth
        }
    )

    controls = widgets.VBox([envelope, extras, smooth, feature], layout=widgets.Layout(width="600px"))
    display(widgets.HBox([controls, plot], layout=widgets.Layout(align_items="center")))
```

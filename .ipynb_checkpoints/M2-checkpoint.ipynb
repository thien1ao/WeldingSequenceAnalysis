```python
import polars as pl
import glob
import numpy as np
import pandas as pd

import plotly.express as px
import matplotlib.pyplot as plt
import warnings

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings('ignore')

%matplotlib inline
```

```python
files = sorted(glob.glob('./data/EV__*.parquet'))

lf = pl.scan_parquet(files)
schema = lf.collect_schema()

lf.head(10).collect()
```

```python
EIN = "[72_1]S7-Schweißmaschine - Schweißen EIN (M215_0)"

df_seq = lf.select("Time", EIN).drop_nulls(subset=["Time", EIN]) # not sure drop_nulls() or drop_nulls(subset=...)
df_seq = df_seq.with_columns(seq_id=(pl.col(EIN) & ~pl.col(EIN).shift(fill_value=False)).cum_sum() - 1)
df_seq = df_seq.filter(pl.col(EIN))
df_seq = df_seq.group_by("seq_id").agg(start=pl.col("Time").min(), end=pl.col("Time").max(), n=pl.len())
df_seq = df_seq.sort("seq_id")

df_seq.head(10).collect(engine="streaming")
```

```python
# keep only numerical columns for standartisation

features = [col for col, dtype in schema.items() if col not in ("Time", EIN) and dtype.is_numeric()]

df_features = lf.select("Time", EIN, *features).drop_nulls(subset=["Time", EIN])
df_features = df_features.with_columns(seq_id=(pl.col(EIN) & ~pl.col(EIN).shift(fill_value=False)).cum_sum() - 1)
df_features = df_features.filter(pl.col(EIN)).select("Time", "seq_id", *features)


df_features.slice(18190, 10).collect(engine="streaming")
```

```python
means = df_features.select([pl.col(col).mean().alias(col) for col in features]).collect(engine="streaming")
stds = df_features.select([pl.col(col).std().alias(col) for col in features]).collect(engine="streaming")

means = np.array([means[col][0] for col in features])
stds = np.array([stds[col][0] for col in features])

summary = pl.DataFrame({
    "feature": features,
    "mean": means,
    "std": stds
})

# check for features with zero standart deviation(constant)

const_features = (
    summary
    .filter(pl.col("std") == 0)
    .get_column("feature")
    .to_list()
)

features = [col for col in features if col not in const_features]
```

```python
means = df_features.select([pl.col(col).mean().alias(col) for col in features]).collect(engine="streaming")
stds = df_features.select([pl.col(col).std().alias(col) for col in features]).collect(engine="streaming")

means = np.array([means[col][0] for col in features])
stds = np.array([stds[col][0] for col in features])

N_SAMPLES = 100

seqs = df_seq.collect(engine="streaming").sort("seq_id")
X = np.empty((len(seqs), len(features) * N_SAMPLES), dtype=np.float32)
seq_ids = seqs["seq_id"].to_list()

norm = [((pl.col(col) - mean) / std).cast(pl.Float64).alias(col) for col, mean, std in zip(features, means, stds)]

for i, row in enumerate(seqs.iter_rows(named=True)):
    seq = (
        lf
        .filter(pl.col("Time").is_between(row["start"], row["end"]))
        .filter(pl.col(EIN))
        .select("Time", *features)
        .sort("Time")
        .with_columns(norm)
        .collect(engine="streaming")
    )

    t = (seq["Time"] - seq["Time"][0]).dt.total_seconds().to_numpy()
    t_lin = np.linspace(0, t[-1], N_SAMPLES)

    for j, feature in enumerate(features):
        y = seq[feature].to_numpy()
        X[i, j*N_SAMPLES:(j+1)*N_SAMPLES] = np.interp(t_lin, t, y)
```

```python
cols = [f"{feature}_t_{i}" for feature in features for i in range(N_SAMPLES)]
df_X = pd.DataFrame(X, index=seq_ids, columns=cols)
df_X.iloc[:10, ::N_SAMPLES]
```

```python
# start with 2 components

pca = PCA(n_components=2)
X_PCA = pca.fit_transform(X)

df_pca = pd.DataFrame(
    X_PCA,
    columns=["PC1", "PC2"],
    index=df_X.index
)

plt.scatter(X_PCA[:, 0], X_PCA[:, 1], color="darkorchid")

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.grid(True, alpha=0.4)
plt.tight_layout()
```

```python
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X)

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], color="darkorchid")

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")
```

```python
pca_full = PCA()
pca_full.fit(X)

cum_var = pca_full.explained_variance_ratio_.cumsum()

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(cum_var) + 1), cum_var, color="darkorchid")

plt.axhline(y=0.9, linestyle="--", label="90% variance", color="black")
plt.axhline(y=0.95, linestyle="--", label="95% variance", color="orange")
plt.axhline(y=0.99, linestyle="--", label="99% variance", color="green")
plt.xlabel("N")
plt.ylabel("CEV")
plt.grid(True)
plt.legend()
plt.show()
```

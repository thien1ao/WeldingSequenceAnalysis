# Bericht

## Vorgehen

1. **Daten laden und Sequenzen bilden.** Parquet-Dateien `EV__*.parquet` mit Polars geladen. Eine Schweißsequenz ist ein zusammenhängender Abschnitt, in dem das Signal „Schweißen EIN" aktiv ist. Jede Sequenz bekommt eine `seq_id`.
2. **Feature-Auswahl und Standardisierung.** Nur numerische Spalten verwendet, konstante Features (Standardabweichung = 0) entfernt. Die übrigen Features standardisiert: (x − Mittelwert) / Standardabweichung.
3. **Lineare Interpolation.** Die Sequenzen haben unterschiedliche Längen. Jedes Feature jeder Sequenz wurde linear auf N = 300 gleichmäßige Zeitpunkte interpoliert (NaN-Werte ausgelassen, Features ohne Messwerte auf 0). Ergebnis: Matrix mit einer Zeile pro Sequenz und `Features × 300` Spalten.
4. **PCA.** Projektion auf 2 Hauptkomponenten zur Visualisierung. Danach kumulierte erklärte Varianz bestimmt; für das Clustering die ersten n = [...] Komponenten behalten (mindestens 95 % der Varianz).
5. **Clustering.** Clusteranzahl mit der Elbow-Methode (Inertia) bestimmt, danach K-Means für k = 2, 3, 4 (50 Initialisierungen).
6. **Vergleich einzelner Sequenzen.** Im PCA-Raum drei Paare gesucht: nächste, am weitesten entfernte und Ausreißer (größter mittlerer Abstand zu den 4 nächsten Nachbarn, plus dessen nächster Nachbar). Für diese Paare ein Feature im Original und interpoliert geplottet.

## Diskussion der Plots

- **PCA-Scatterplot (PC1/PC2).** Zeigt, ob die Sequenzen Gruppen bilden oder ein Kontinuum. Einzelne Punkte weit außerhalb sind Ausreißer-Kandidaten. Zwei Komponenten erklären nur einen Teil der Varianz, die Trennung ist daher nur eine Näherung. Beobachtung: [...]
- **Kumulierte Varianz.** Je flacher der Anstieg, desto weniger lässt sich die Dimension reduzieren. Für 95 % werden n = [...] Komponenten benötigt (ursprünglich `Features × 300` Dimensionen).
- **Elbow-Plot.** Knick bei k ≈ [...]. Er ist oft nicht eindeutig, daher zusätzlich k = 2, 3, 4 verglichen.
- **Cluster im PCA-Raum.** Die Cluster wurden auf n Komponenten berechnet, aber nur in PC1/PC2 dargestellt. Überlappungen im Plot bedeuten nicht zwingend Überlappung im vollen Raum. Beobachtung: [...]
- **Nächstes, entferntestes und Ausreißer-Paar.** Die Markierungen im PCA-Plot zeigen die Lage der Paare. In den Zeitreihen sollten sich nahe Sequenzen ähnlich verhalten und entfernte deutlich unterscheiden. Beobachtung: [...]
- **Original vs. interpoliert.**
  - Die Form der Kurve bleibt erhalten, die lineare Interpolation glättet nur schnelle Spitzen.
  - Die Zeitachse wird gestreckt bzw. gestaucht, die absolute Dauer geht verloren. Für den Formvergleich gewollt, kann aber Unterschiede in der Dauer verdecken.
  - Interpolierte Werte sind standardisiert, die Originale nicht; die Y-Achsen sind daher nicht direkt vergleichbar.

## Einschränkungen

- Die lineare Interpolation auf gleiche Länge ignoriert die Dauer der Sequenz. Eine Alternative wäre Dynamic Time Warping.
- Die Standardisierung nutzt Mittelwert und Standardabweichung über alle Daten; Features mit wenigen Messwerten oder vielen NaN sind weniger zuverlässig.

# Bericht Meilenstein 3: Clustering von Schweißsequenzen

## K-Means

<img src="assets/elbow_silhouette.png" width="600">

Um die Anzahl der Cluster zu bestimmen, wurden die Elbow-Methode und der Silhouette-Score verwendet. Die Inertia zeigt einen Knick bei \(k = 2\), und auch der Silhouette-Score ist dort mit 0,32 am höchsten. Bei \(k = 3\) ist der Wert mit 0,29 nur etwas niedriger. Daher kommen vor allem zwei oder drei Cluster infrage. Bei mehr als drei Clustern wird die Trennung zwischen den Gruppen deutlich schlechter.

<img src="assets/kmeans_k234.png" width="600">

Die PCA-Darstellung zeigt, dass sich bei \(k = 3\) drei Gruppen erkennen lassen. Dabei wird die rechte Gruppe weiter unterteilt, während die linke Gruppe weitgehend zusammenbleibt. Bei \(k = 4\) ist dagegen keine klare Trennung der zusätzlich entstehenden Gruppen erkennbar. Zusammen mit den Ergebnissen der beiden Metriken spricht dies für die Wahl von \(k = 3\).

## DBSCAN und der Einfluss von eps und min_samples

<img src="assets/dbscan_kdistance.png" width="700">

<img src="assets/dbscan_matrizen.png" width="800">

Bei DBSCAN wurde `eps` zunächst mithilfe des k-Distanz-Plots abgeschätzt und anschließend gemeinsam mit `min_samples` untersucht. Die drei Matrizen zeigen die Anzahl der gefundenen Cluster, den Anteil der Rauschpunkte und den Silhouette-Score. Dabei wird deutlich, dass kleine `eps`-Werte zu vielen kleinen Clustern und einem hohen Rauschanteil führen. Bei größeren Werten verschmelzen die Gruppen dagegen zunehmend. Drei Cluster mit einem geringen Rauschanteil entstehen nur in einem kleinen Bereich um `eps` = 90–97 und `min_samples` = 5–15. Das zeigt, dass DBSCAN stark von der Wahl seiner Parameter abhängt und damit empfindlicher als K-Means ist. Außerdem ist zu erkennen, dass höhere Silhouette-Werte häufig mit einem höheren Rauschanteil einhergehen.

## Visualisierung und Vergleich der Verfahren

<img src="assets/methodenvergleich.png" width="900">

Die Abbildung zeigt die Ergebnisse von K-Means und fünf DBSCAN-Varianten mit jeweils drei Clustern in der PC1/PC2-Ebene. Alle Verfahren finden im Wesentlichen dieselben drei Gruppen. Der wichtigste Unterschied besteht darin, dass K-Means alle Sequenzen zuordnet, während DBSCAN je nach Parametern 4 bis 18 % als Rauschen markiert. Diese Punkte liegen hauptsächlich zwischen den beiden rechten Clustern, wo die Gruppen ineinander übergehen. Bei `eps` = 97 und `min_samples` = 5 werden diese beiden Gruppen dagegen zusammengefasst.

## Untersuchung der Cluster

<img src="assets/zeitreihen_vorschub_strom.png" width="900">

Um zu prüfen, ob sich die Cluster auch in den ursprünglichen Zeitreihen widerspiegeln, wurde pro Cluster die Sequenz ausgewählt, die dem Clusterzentrum am nächsten liegt. Verglichen wurden die`C-Wagen-Position` und der `Schweißstrom`. Aufgrund seiner starken Oszillation wird der `Schweißstrom` zur besseren Visualisierung als Min/Max-Hüllkurve dargestellt.

Zusätzlich sind einige Ausreißer eingezeichnet. Die braune gepunktete Linie („Cluster 0 (am weitesten)“) zeigt die Sequenz aus Cluster 0 mit dem größten Abstand zum Clusterzentrum. Die rote gestrichelte Linie (Ausreißer 0) und die violette strichpunktierte Linie (Ausreißer 1) zeigen die beiden DBSCAN-Ausreißer mit dem größten Abstand zum Clusterzentrum. Alle drei Sequenzen weichen in ihrer Dauer, Bewegungsgeschwindigkeit oder ihrem Stromniveau von den typischen Verläufen ihrer Cluster ab.

## Dynamische Features und die Einspeisung

<img src="assets/einspeisung_spannung_strom.png" width="900">

Unter den Features mit der größten zeitlichen Streuung befinden sich vier aus der Einspeisung. Die drei Spannungsfeatures (`Einspeisung L1/L2/L3 Spannung Istwert`) zeigen eine schnelle Schwingung, die durch die Interpolation auf 300 Punkte vermutlich eher zufällig abgetastet wird. Dadurch liefern sie kaum zusätzliche Informationen für die Clusterbildung und könnten sogar Rauschen einbringen. Anders verhält es sich mit `Einspeisung L1 Strom Istwert` (rechts), dessen Niveau sich zwischen den Clustern unterscheidet.

## Zusammenfassung

Die drei Cluster lassen sich anhand ihrer Dauer, Vorschubgeschwindigkeit und ihres Stromniveaus gut beschreiben. K-Means ordnet alle Sequenzen genau einer Gruppe zu, während DBSCAN einige Sequenzen zwischen den Clustern als Rauschen markiert. Beide Verfahren finden eine ähnliche Gruppierung, das Ergebnis von DBSCAN hängt aber stärker von den gewählten Parametern ab. Deshalb wurde K-Means für die weitere Analyse bevorzugt und DBSCAN ergänzend betrachtet.
